```python
import os
from PIL import Image

# 예제 데이터
sentiment = "기쁨"  # 한글 문자열
img_path = "1.jpg"

# 이미지 경로 생성
image_file_path = f"original_data/EMOIMG_{sentiment}_SAMPLE/{img_path}"

# 파일 경로 출력
print("이미지 파일 경로:", image_file_path)

# 파일이 존재하는지 확인
if os.path.exists(image_file_path):
    print("파일이 존재합니다.")
    # 이미지 열기 및 RGB로 변환
    img = Image.open(image_file_path).convert("RGB")
    # 이미지 확인 (선택 사항)
    img.show()
else:
    print("파일이 존재하지 않습니다.")
```

```python
import os

sentiments = ["기쁨", "당황", "분노", "상처"]

for sentiment in sentiments:
    # 디렉토리 경로
    directory = f'original_data/EMOIMG_{sentiment}_SAMPLE/'

    # 변경할 파일 이름들을 리스트로 준비
    file_names = os.listdir(f"original_data/EMOIMG_{sentiment}_SAMPLE")


    # 파일 이름을 순회하며 순차적으로 변경
    for i, file_name in enumerate(file_names):
        try:
            # 기존 파일 경로
            old_path = os.path.join(directory, file_name)
            
            # 새 파일 이름
            new_file_name = "s"+str(i)+".jpg"
            new_path = os.path.join(directory, new_file_name)
            
            # 파일 이름 변경
            os.rename(old_path, new_path)
            
            print(f"파일 이름을 변경했습니다: {file_name} -> {new_file_name}")
        except FileNotFoundError:
            print(f"파일을 찾을 수 없습니다: {file_name}")
        except Exception as e:
            print(f"오류 발생: {e}")
```
